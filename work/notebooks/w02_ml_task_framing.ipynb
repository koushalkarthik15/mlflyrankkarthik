## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (like 'if age > 180 days and impressions dropping') is a good baseline, but it is rigid. ML beats a fixed rule because it can weigh dozens of weak signals simultaneously—such as the interaction between CTR, average position, content type, and engagement rate—to spot non-obvious decay risks that a simple if-statement would miss.
