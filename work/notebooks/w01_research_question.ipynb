## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# 1. Load the starter data
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
total_pages = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
low_ctr_pages = len(df[(df['impressions_90d'] > 1000) & (df['ctr'] < 1.0)])

print(f"Total pages in the dataset: {total_pages:,}")
print(f"Pages actively declining in traffic: {declining_pages:,} ({(declining_pages/total_pages)*100:.1f}%)")
print(f"Pages with high impressions (>1000) but awful CTR (<1%): {low_ctr_pages:,}")
print("\nConclusion: With thousands of declining or underperforming pages, human editors cannot review them all manually. We need a scoring system to rank the most urgent ones first.")
```
